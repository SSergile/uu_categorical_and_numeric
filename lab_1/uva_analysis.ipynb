Analyze foreign giving to UVA. Compare UVA to appropriate peer institutions and the market overall. Propose a strategy for finding new sources of foreign money and provide a quantitative analysis.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('ForeignGifts_edu.csv',low_memory=False)
df.head()
```

```python
u_totals =df.groupby('Institution Name')['Foreign Gift Amount'].sum()
median = u_totals.median()
print(median)
```

```python
competitors = [
    'University of Virginia',
    'University of North Carolina - Chapel Hill',
    'Duke University',
    'University of Notre Dame',
    'Virginia Polytechnic Institute & State University',
]
gift_totals = (
    df[df['Institution Name'].isin(competitors)]
    .groupby('Institution Name')
    ['Foreign Gift Amount'].sum()
    .sort_values(ascending=False)
)
print(gift_totals)
```

```python
gift_counts = (
    df[df['Institution Name'].isin(competitors)]
    .groupby('Institution Name')
    .size()
    .sort_values(ascending=False)
)

print(gift_counts)
```

While UVA ($22,189,238) is well ahead of the median college in terms of foreign gifts received by school ($6,486,791.5), when compared to its peer institutions in the top of the ACC, they are not doing well, receiving not only less gift dollars, but less total gifts as well.

```python
average_gift = (
    df[df['Institution Name'].isin(competitors)]
    .groupby('Institution Name')['Foreign Gift Amount']
    .mean()
    .sort_values(ascending=False)
)

print(average_gift)
```

This tells us that UVA has an average higher than UNC, which means they get more money than UVA because they get significantly more gifts. UVA should focus their efforts on acquiring more gifts.

```python
uva = df[df['Institution Name'] == 'University of Virginia']

country_counts = uva['Country of Giftor'].value_counts()

print(country_counts)
```

```python
country_totals = uva.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=False)

print(country_totals)
```

England has given UVA the most gifts. Despite this, Bangladesh has given UVA the most money

```python

uva_countries = set(df[df['Institution Name'] == 'University of Virginia']['Country of Giftor'])

competitor_country_sets = [
    set(df[df['Institution Name'] == school]['Country of Giftor'])
    for school in competitors
]

countries_in_all_competitors = set.intersection(*competitor_country_sets)

missing_from_uva = countries_in_all_competitors - uva_countries

print(missing_from_uva)
```
