```python
import numpy as np
import pandas as pd
from scipy.stats import uniform
import matplotlib.pyplot as plt
```

**Question 1**

```python
# set up grid of equally spaced values between 0 and 1
k = 10
grid = np.arange(1, k+1)/k
print(grid)
#draw a point from the grid at random with equal probability
point = np.random.choice(grid)
#for every k, repeat 5000 times, plot the ecdf of the draws
n = 5000
entry = np.random.choice(grid, n)
x = np.sort(entry)
y = np.arange(1, n+1)/n
plt.step(x, y, where ='post')


```

As you increase k, the ECDF of the draws gets closer and closer to a straight line. The jumps get smaller and smaller.

**Question 2**

```python
n = 2
b = 5000
results = np.empty(b)
for i in range(b):
    uniform = np.random.uniform(-np.sqrt(3),np.sqrt(3), n)
    results[i] = np.mean(uniform) * np.sqrt(n)

x = np.sort(results)
y = np.arange(1, b+1)/b
plt.step(x,y, where = 'post')
```

As N increases, the graph approaches a cumulative distribution function (CDF)

**Question 3**

```python
n = 100
p = 1/10
tails = 0
heads = 1
rand = [tails,heads]
b = 5000
results = np.empty(b)
for i in range(b):
    result = np.empty(n)
    for m in range(n):
        x = np.random.choice(rand)
        result[m] = x
    count = (result == heads).sum()
    entry = (count-(n*p))/(np.sqrt(n*p*(1-p)))
    results[i] = entry

x = np.sort(results)
y = np.arange(1, b+1)/b
plt.step(x,y, where = 'post')
```

As n increases, the graph approaches a cumulative distribution function.

When p decreases, the graph shape stays the same, but the x values get bigger.
When p increases, the graph shape stays the same, but the x values get smaller.

**Question 5**

```python
k = 20
n = 5000
for m in range(1,k+1):
    results = np.empty(n)
    for i in range(n):
        dist = np.random.uniform(0,1,k)
        sorted = np.sort(dist)
        results[i] = sorted[m-1]
    x = np.sort(results)
    y = np.arange(1, n+1)/n
    plt.step(x,y, where = 'post')
```

This process generates 20 different cumulative distribution functions. Each distribution depends on the rank, because as rank (k) increases, the CDFs shift to the right, away from 0 and closer to 1. The distribution halfway between 1 and k is the one that is closest to a cdf, because it's the most symmetric. CDFs closer to 1 or k have either a shortened head or tail because they are bound by 0 and 1.
